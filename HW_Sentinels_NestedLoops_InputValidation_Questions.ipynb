## Task 1 — Sentinel stats (≈15 min)

Write a program that repeatedly asks the user for **integers between 0 and 100**.  
Use a **sentinel**: the user types `done` (any case) to finish.

**Requirements**
- Ignore blank inputs (reprompt).
- If the user types something that's not an integer in range, print a friendly message and reprompt.
- When the user finishes, print: **count**, **sum**, **min**, **max**, and **mean** (to 2 decimals) using f-string: `{num:.2f}`.
- If no valid numbers were entered, print `"No data."`

````python

num_list=[]
print("Please enter a number between 0 and 100, when ready to quit enter 'done'.")

while True:
  answer= input("Please enter a number").strip() #so we can always recognize the users 'done'

  if answer.lower()=="done":
    break
  if answer== "":
    continue
  if (int(answer)) <0:
    print("Your number must be between 0 and 100")
    continue
  if (int(answer)) >100:
    print("Your number must be between 0 and 100")
    continue

  num_list.append(int(answer))
  print(num_list)
count=len(num_list)
total=sum(num_list)
minimum= min(num_list)
maximum=max(num_list)
print(f"Count : {count}, Sum : {total}, Min : {minimum}, Max : {maximum}, Mean : {total/count}")

````
