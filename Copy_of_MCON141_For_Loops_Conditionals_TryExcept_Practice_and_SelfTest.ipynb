## 9. `try/except` Inside a `for` Loop

Each item is tried separately. Bad data does not have to crash the whole loop.

**Before running:** trace the values by hand.

````python
entries = ["12", "7", "hello", "20"] #enteries for the following loop

for entry in entries: #establishes loop
    try:#a regular try/except ValueError statement
        number = int(entry)#entry must be an integer, then it will print number, 'is a valid integer'
        print(number, "is a valid integer")
    except ValueError:
        print(entry, "is NOT an integer")#if not an integer it is an incorrect entry and it will print entry, 'is NOT an integer'
````

## 10. `try/except` + Conditional

First convert safely; then make a decision about the converted number.

**Before running:** trace the values by hand.

````python
entries = ["4", "11", "cat", "18"]#entries for the following loop

for entry in entries:#establishes loop
    try:#a regular try/ Except ValueError conditional statement
        number = int(entry) #entry should be an integer
        if number >= 10:#if the number is greater then or equal to 10 it will print number,'is at least 10'
            print(number, "is at least 10")
        else:#if the number is any other value it will print number,'is less than 10'
            print(number, "is less than 10")
    except ValueError:
        print(entry, "cannot be converted")#any other entry is incorrect and will print entry, 'cannot be converted'
````

## 11. `try/except` + `continue`

If conversion fails, skip the rest of that iteration.

**Before running:** trace the values by hand.

````python
entries = ["5", "oops", "8", "bad", "12"] #entries for the loop

for entry in entries:#establishes the loop
    try:#try/except/continue=if except ValueError, it will skip that entry entirely
        number = int(entry)#entry should be an integer
    except ValueError:#if anythinthing but an integer will print 'Skipping', entry
        print("Skipping:", entry)
        continue #will skip and continue to next value in the loop

    print(number, "squared is", number ** 2)
````

## 12. Putting It Together

Convert each value. Skip bad input. Ignore negatives. Stop at 100. Classify the remaining numbers as even or odd.

**Before running:** trace the values by hand.

````python
entries = ["8", "cat", "-3", "11", "24", "100", "6"]#entries for loop

for entry in entries:#establishes loop
    try:#try to convert each value
        number = int(entry)#value should be an integer
    except ValueError:#if not an integer and therefore unable to convert, print 'bad input:', entry
        print("Bad input:", entry)
        continue #continue to next entry

    if number < 0:#if the number is less than 0
        print("Ignoring negative:", number)#print 'ignoring negative:', number
        continue

    if number == 100:#if number is equal to 100
        print("100 found — stopping")#print '100 found - stopping'
        break#stop the loop

    if number % 2 == 0:#if number is modulo==0, (divisable by 2) print number, 'even'
        print(number, "even")
    else:
        print(number, "odd")#all others print number, 'odd'
````
