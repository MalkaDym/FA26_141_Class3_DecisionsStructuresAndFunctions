## Reference sheet — Read before beginning

These are **small syntax fragments**, not complete programs. You must decide which ideas to use, where to place them, and how to combine them.

**Exception handling**
- `try:` begins the statements that might fail.
- `except ValueError:` catches an invalid numeric conversion or a deliberately raised `ValueError`.
- `raise ValueError("message")` deliberately signals an invalid value.
- `int(text)` converts whole-number text; `float(text)` converts decimal-number text.
- `else:` after `except` runs only when no exception occurs; `finally:` runs regardless of whether an exception occurs. Use these only if helpful.

**Sentinel-controlled repetition**
- A *sentinel* is a special input that ends repetition, such as `done`, `0`, or `-1`.
- `while answer != "done":` is an example of a continuation condition.
- Decide where to read the first value, where to check for the sentinel, and where to read the next value.
- A counter can be updated using `count += 1`; a total using `total += number`.
- For a numeric sentinel, decide how to handle invalid text without accidentally ending the loop.

**Nested loops**
- `for x in range(1, 4):` iterates over `1`, `2`, `3`; the stop value is excluded.
- `while x <= limit:` repeats as long as its condition is true.
- The *inner* loop completes before the *outer* loop advances.
- For `while` loops, identify which variables must change and which must be reset.
- `print(value, end=" ")` keeps output on the same line; `print()` begins a new line.
- `break` exits the nearest enclosing loop; `continue` skips to its next iteration. Use them thoughtfully.

**Boolean logic**
- `and` requires both operands to be true; `or` requires at least one; `not` reverses a Boolean.
- Precedence is `not`, then `and`, then `or`. Use parentheses to make your intent explicit.
- De Morgan's laws: `not (A and B)` equals `(not A) or (not B)`; `not (A or B)` equals `(not A) and (not B)`.
- Comparison operators: `==`, `!=`, `<`, `<=`, `>`, `>=`.
- `number % 2 == 0` tests whether an integer is even.

**Strings and lists — limited tools**
- `input("Prompt: ")` returns a string.
- `answer.strip().lower()` removes surrounding spaces and normalizes capitalization.
- `answer == "done"` compares a string with a sentinel.
- `for letter in word:` visits each character of a string, one at a time. You do not need to index the string.
- `len(word)` counts characters in a string; `len(items)` counts items in a list.
- `items[0]` accesses the first list item (indexing begins at zero).
- `for item in items:` visits each item in a list.
- If you choose to collect values in a list, start with `values = []`, then use `values.append(number)` to add a value. **Appending is not required in this lab.**

```python
# YOUR CODE HERE

count= 0
sum= 0

while True:

  number = input("Please enter a positive number. When ready to quit write 'done' ").strip().lower()

  if number=="done":
    break

  try:
    number=int(number)

    if number > 0:
      count += 1
      sum += number

    if number < 0:
      raise ValueError("The number you enter must be positive.")

  except ValueError as e:
    ("Please try again", e)



print(f"You have entered {count} positive numbers that totals to {sum}")
```
