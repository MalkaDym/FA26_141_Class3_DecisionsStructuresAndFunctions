```python
name= input("Please enter your name: ") #asks the user their name
age= input("How old are you? ") #asks the user their age
vip= input("Do you have VIP access? y/n").strip().lower()#asks the user their VIP status
admission_fee=0 #admission accumulator
try:
    age=int(age) #converts all age inputs to an integer
    age > 0
    vip == 'y' or 'n' #only accepts y/n
    name == str #only accepts a string as a name
except ValueError:

    print("Incorrect. Please re-enter your information.") #raises this error if the above criteria in the try statement are not met

while True:
  if vip=='y':#if they are a vip member, admission is free
    admission_fee=0
    print("Welcome! Admission is free!!" )
    break
  elif age < 8: #if they are younger than 8, they must be accompanied by an adult
    print("You must be accompanied by an adult.")
    break
  elif age >= 8 and age <= 17: #between ages 8 and 17 they are charged a $5 admission fee
    admission_fee=5
    print("Welcome! You will be charged a $5 admission fee.")
    break
  else: #above 17 (all else) are charged $10
    admission_fee=10
    print("Welcome! Your admission fee is $10.")
    break
```

```python
total = 0

while True:
    purchase= input("How many tickets? Write 'done' to quit").strip().lower() #asks the user how many tickets they have, gives an option to end the loop with 'done' as a sentinel to end while loop
    if purchase == 'done':
      break

    try:
        purchase=int(purchase)

        if purchase < 1 and purchase > 20: #does not allow user to purchase more than 20 tickets
          raise ValueError("Tickets must be between 1-20")


        if purchase > 0 and purchase <= 20:
            total += purchase
            print("Accepted: ", purchase)
        if total >= 20: #will stop the purchase at 20 tickets
          break

    except ValueError:
        print("Please enter a whole number.")



print(f"Your total price for your ticket purchase comes to ${total*2}") #total amount of tickets *2 since each ticket is $2
```

```python
tickets= int(input("How many tickets do you have?"))#user must enter their amount of tickets
correct_prize=0 #correct prize accumulator (10pts)
incorrect_prize=0 #incorrect prize accumulator (2pts)

for guess in range(tickets):
  guess= input("Guess the number between 1 and 5. Write 'done' when ready to quit.") #asks user for number guess, or option to use sentinel to end the for loop
  tickets -= 1 #subtracts from amount of tickets user has for every turn
  if guess== 'done':
      break
  if tickets <= 0:#breaks when tickets run out
      break

  try:
    guess=int(guess)#converts guess into integer

    if guess > 5 or guess < 1: #only accepts a number between 1 and 5
      raise ValueError("Please guess again")
      continue

  except ValueError:
    print("Invalid. Please guess again")
    continue

  if guess == 3: #if guess 3 you get the full 10 points
    print("Correct! You earned 10 points.")
    correct_prize += 10 #+10 to the point count
  else: #if guess anything in between 1-5 (except 3) you get 2 points
    print("Nice try, your guess was incorrect but you still gained 2 points!")
    incorrect_prize += 2 #+2 to the point count

print(f"Thank you for playing! Your total score is {correct_prize + incorrect_prize}")
```

```python
position= 0
row=0

while row < 5:# outer loop= inner loop goes first, this one goes after a full round of the inner loop
  col=0 #col remains at 0 until a full loop of rows pass through
  while col < 5:#inner loop= completed first
    if row==col or row+col==4:
      print("*", end=" ") #if the rows are equal or the columns equal 4 it prints a star
    else:
      print("-",end=" ") #if not it prints a -
    position +=1
    col+=1
  print()
  row+=1
print("Position placement", position) #names the position placement
```
