```python
#get temperature

while True:

    #removed conversion to int when asking for sentinal thats a string
    temp = input("Please enter a temperature in farenheit : \
     accepting values in the range of -20 to 120 \
     or write 'done' ").strip() #has to be stripped so its normalized and could be recognized

    if temp.lower() == "done": #has to be lowered so its normalized and could be recognized
      print("I see that you are done ..Exiting...")
      break

    else:
      try:
        temp = int(temp)
        if temp  < -20 or temp > 120:
          raise ValueError("This was a bad temp choice-- try again")

      except ValueError as e:
        print("enter a number please", e)
        continue
      else:
        print(f"Great tempurature I like it! {temp}")
```
