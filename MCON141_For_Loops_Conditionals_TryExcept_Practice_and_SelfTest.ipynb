## 2. Decision Inside a `for` Loop

For each number, decide whether it is even or odd.

**Before running:** trace the values by hand.

````python
for number in range(1, 7): #establishes the loop using 'for' 'in' and by writing (1,7) it will print 1,2,3,4,5,6
#with every cycle in the loop it will test the number against the if/else statement and print the corrosponding response
    if number % 2 == 0:
        print(number, "is even")#if the number is divisible by 2 then it will print 'is even'
    else:
        print(number, "is odd") #if not it will print odd
````

## 3. `if / elif / else` Inside a Loop

Classify each score.

**Before running:** trace the values by hand.

````python
scores = [95, 82, 67, 54] #will print the following scores

for score in scores:#establishes loop, the following statements will be run in every cycle of the loop until all the scores have been printed and they function as a regular if/elif/else statement and print the corrosponding mesage for every score
    if score >= 90:
        print(score, "A range")#if greater than 90 will print 'A range'
    elif score >= 80:
        print(score, "B range")#if greater than 80 will print 'B range'
    elif score >= 70:
        print(score, "C range")#if greater then 70 will prin 'C range'
    else:
        print(score, "Below C range")#all else will print 'below C range'
````

## 4. Compound Logic with `and`

Only values from 10 through 20 should be accepted.

**Before running:** trace the values by hand.

````python
values = [4, 10, 15, 20, 25] #will print the following scores

for value in values:#establishes the loop function
    if value >= 10 and value <= 20:#works as a regular if/else statement with 'and' but will determine if the number falls under this category every time it loops around for the next value
        print(value, "is in range")#if is between 10 and 20 will print 'is in range'
    else:
        print(value, "is outside the range")#if otherwise will print 'is outside range'
````

## 5. Compound Logic with `or`

Mark very low or very high temperatures.

**Before running:** trace the values by hand.

````python
temperatures = [20, 45, 70, 95]

for temp in temperatures:#works as a regular if/else statement with 'or' but will determine if the number falls under this category every time it loops around for the next tempurature
    if temp < 32 or temp > 90:
        print(temp, "EXTREME")#if temp is under 32 or above 90 degrees will print temp, 'EXTREME'
    else:
        print(temp, "moderate")#all other temp will print temp, 'moderate'
````

## 6. `continue` with a Conditional

Skip negative numbers but process the others.

**Before running:** trace the values by hand.

````python
numbers = [8, -2, 5, -7, 3]

for number in numbers:
    if number < 0:
        continue
    print("Using:", number)
````

## 7. `break` with a Conditional

Stop as soon as the sentinel value `0` is found.

**Before running:** trace the values by hand.

````python
numbers = [8, 5, 3, 0, 9, 12]

for number in numbers:
    if number == 0:
        print("Stop value found")
        break
    print("Processing:", number)
````

## 8. `enumerate()`

`enumerate()` gives us both a position and a value.

**Before running:** trace the values by hand.

## 9. `try/except` Inside a `for` Loop

Each item is tried separately. Bad data does not have to crash the whole loop.

**Before running:** trace the values by hand.

````python
entries = ["12", "7", "hello", "20"]

for entry in entries:
    try:
        number = int(entry)
        print(number, "is a valid integer")
    except ValueError:
        print(entry, "is NOT an integer")
````

## 10. `try/except` + Conditional

First convert safely; then make a decision about the converted number.

**Before running:** trace the values by hand.

````python
entries = ["4", "11", "cat", "18"]

for entry in entries:
    try:
        number = int(entry)
        if number >= 10:
            print(number, "is at least 10")
        else:
            print(number, "is less than 10")
    except ValueError:
        print(entry, "cannot be converted")
````

## 11. `try/except` + `continue`

If conversion fails, skip the rest of that iteration.

**Before running:** trace the values by hand.

````python
entries = ["5", "oops", "8", "bad", "12"]

for entry in entries:
    try:
        number = int(entry)
    except ValueError:
        print("Skipping:", entry)
        continue

    print(number, "squared is", number ** 2)
````

## 12. Putting It Together

Convert each value. Skip bad input. Ignore negatives. Stop at 100. Classify the remaining numbers as even or odd.

**Before running:** trace the values by hand.

````python
entries = ["8", "cat", "-3", "11", "24", "100", "6"]

for entry in entries:
    try:
        number = int(entry)
    except ValueError:
        print("Bad input:", entry)
        continue

    if number < 0:
        print("Ignoring negative:", number)
        continue

    if number == 100:
        print("100 found — stopping")
        break

    if number % 2 == 0:
        print(number, "even")
    else:
        print(number, "odd")
````
